```python
import matplotlib.pyplot as plt
import os 

Data_path ="datasets/train"
for image in os.listdir(Data_path)[:2]:
    img = plt.imread(os.path.join(Data_path, image))
    plt.imshow(img)
    plt.show()
```

```python
list0=[]
for image in os.listdir(Data_path):
    img=plt.imread(os.path.join(Data_path,image))
    list0.append(img.mean(axis=(0,1)))

mean=0
for i in list0:
    mean+=i
print(mean/len(list0))

plt.plot(list0)
```
